```python
import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud
```

```python
real = pd.read_csv("True.csv")
fake = pd.read_csv("Fake.csv")
```

```python
real_words_text = " ".join(real["text"].head(2000).fillna("").astype(str))
fake_words_text = " ".join(fake["text"].head(2000).fillna("").astype(str))
```

```python
# real news word cloud
real_wordcloud = WordCloud(width=800, height=400, background_color="white", stopwords=WordCloud().stopwords).generate(real_words_text)

plt.figure(figsize=(10, 5))
plt.imshow(real_wordcloud, interpolation="bilinear")
plt.title("Word Cloud - Real News")
plt.axis("off")
plt.show()
```

```python
# fake news word clouud
fake_wordcloud = WordCloud(
    width=800, height=400, background_color="white",stopwords=WordCloud().stopwords).generate(fake_words_text)

plt.figure(figsize=(10, 5))
plt.imshow(fake_wordcloud, interpolation="bilinear")
plt.title("Word Cloud - Fake News")
plt.axis("off")
plt.show()
```

```python
# both word clouds
plt.figure(figsize=(16, 7))

plt.subplot(1, 2, 1)
plt.imshow(real_wordcloud, interpolation="bilinear")
plt.title("Real News", fontsize=16)
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(fake_wordcloud, interpolation="bilinear")
plt.title("Fake News", fontsize=16)
plt.axis("off")

plt.suptitle("Word Frequency Comparison: Real vs Fake News", fontsize=18)
plt.tight_layout()
plt.show()
```
